```python
import pandas as pd

dataset = pd.read_csv("../dataset/cleaned_dataset.csv")
```

```python
dataset.info()
```

```python
dataset["saledate"] = pd.to_datetime(dataset["saledate"], errors = "coerce", utc = True)

dataset["sale_year"] = (dataset["saledate"].dt.year).astype("int64")

dataset["vehicle_age"] = dataset["sale_year"] - dataset["year"]

dataset.loc[(dataset["vehicle_age"] < 0), "vehicle_age"].value_counts().sort_values()
```

```python
# Knowing about older vehicles

dataset["vehicle_age"].value_counts().sort_index().tail(20)
```
