```python
# Checking NULL values
dataset["vin"].isnull().sum()

# Checking overall duplicate rows
dataset.duplicated().sum()

# duplicate "vin"s
dataset["vin"].duplicated().sum()

```

```python
# changing "state" into string
dataset["state"] = dataset["state"].astype("string")
dataset["state"].info()

dataset["state"] = dataset["state"].str.lower().str.strip()

# checking invalid states
invalid_state = dataset["state"][dataset["state"].str.len() > 2]

# checking invalid states count
invalid_state.value_counts().sum()

# checking valid states
valid_state = dataset["state"][dataset["state"].str.len() <= 2]

# Total valid states
valid_state.unique().size

dataset["state"].unique().size
```

```python
dataset.info()
dataset.columns
```

```python
# changing datatypes

for col in ['make', 'model', 'trim', 'body', 'transmission', 'color', 'interior', 'seller']:
    dataset[col] = dataset[col].astype("string")
    dataset[col] = dataset[col].str.strip().str.lower()

dataset["saledate"] = pd.to_datetime(dataset["saledate"], errors = "coerce")
```
